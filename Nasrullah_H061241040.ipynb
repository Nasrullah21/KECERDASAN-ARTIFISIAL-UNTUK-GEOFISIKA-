### **1️. STUDI KASUS GEOLISTRIK**

### **Deskripsi**

Pada studi kasus ini, kita menggunakan hasil analisis **geolistrik** berupa penampang resistivitas 2D.

Fokusnya adalah melihat pola resistivitas dan menginterpretasikan kondisi lapisan atau material yang ada di bawah permukaan.

### **Role**

> Anda berperan sebagai pendamping dalam membantu memahami dan menjelaskan hasil analisis geolistrik.

### **Task**

> Bantu kita menginterpretasikan hasil geolistrik untuk melihat kondisi bawah permukaan dan kemungkinan lapisan atau material yang ada.

### **Context**

> Geolistrik digunakan untuk melihat perbedaan nilai resistivitas di bawah permukaan. Perbedaan nilai tersebut dapat memberikan gambaran mengenai perbedaan material atau kondisi geologi bawah permukaan.

### **Input**

> Kita akan memberikan penampang resistivitas 2D, nilai resistivitas, kedalaman, serta informasi geologi jika tersedia.

### **Constraints**

> Jangan mengolah ulang data dan jangan membuat nilai baru. Interpretasi harus berdasarkan hasil yang diberikan dan harus membedakan antara hasil yang terlihat dengan interpretasi.

### **Output**

> Jelaskan pola resistivitas, pembagian lapisan, kemungkinan material penyusun, serta zona yang memiliki perubahan atau anomali resistivitas.

### **Quality Criteria**

> Interpretasi harus sesuai dengan penampang dan nilai resistivitas yang diberikan serta menggunakan bahasa yang mudah dipahami.

### **Verification**

> Periksa kembali apakah interpretasi sudah sesuai dengan penampang resistivitas dan informasi geologi yang tersedia.

## **2️. STUDI KASUS DInSAR**

### **Deskripsi**

Pada studi kasus ini, kita menggunakan hasil analisis **Differential Interferometric Synthetic Aperture Radar (DInSAR)**.

Fokusnya adalah melihat apakah terdapat deformasi permukaan dan bagaimana pola deformasi tersebut.

### **Role**

> Anda berperan sebagai pendamping dalam membantu memahami dan menjelaskan hasil analisis DInSAR.

### **Task**

> Bantu kita melihat dan menginterpretasikan pola deformasi permukaan dari hasil DInSAR.

### **Context**

> DInSAR digunakan untuk melihat perubahan permukaan bumi dari dua waktu pengamatan. Hasilnya dapat digunakan untuk melihat apakah suatu wilayah mengalami perubahan atau deformasi.

### **Input**

> Kita akan memberikan interferogram, peta displacement, nilai displacement, serta informasi gempa atau kondisi geologi jika tersedia.

### **Constraints**

> Jangan mengolah ulang data dan jangan membuat nilai baru. Jangan langsung menganggap semua perubahan warna sebagai deformasi.

### **Output**

> Jelaskan wilayah yang mengalami deformasi, besar dan arah perubahan jika informasinya tersedia, serta kemungkinan penyebab dari deformasi tersebut.

### **Quality Criteria**

> Interpretasi harus berdasarkan hasil yang diberikan dan harus membedakan mana hasil pengamatan dan mana yang merupakan dugaan atau interpretasi.

### **Verification**

> Periksa hasil interpretasi dengan informasi gempa, GNSS, atau kondisi geologi jika data tersebut tersedia.





## **3️. STUDI KASUS InSAR TIME SERIES**

### **Deskripsi**

Pada studi kasus ini, hasil InSAR digunakan untuk melihat perubahan deformasi tidak hanya berdasarkan lokasi, tetapi juga berdasarkan **waktu**.

Hasil yang digunakan dapat berupa **velocity map** dan **grafik time series displacement**.


### **Role**

> Anda berperan sebagai pendamping dalam membantu kita memahami hasil analisis InSAR.

### **Task**

> Bantu kita menginterpretasikan perubahan deformasi berdasarkan peta kecepatan dan grafik perubahan terhadap waktu.

### **Context**

> InSAR dapat digunakan untuk melihat perubahan permukaan bumi dari beberapa waktu pengamatan. Dengan hasil tersebut kita dapat melihat bagaimana deformasi berubah secara spasial dan terhadap waktu.

### **Input**

> Kita akan memberikan velocity map, grafik time series, nilai displacement, dan lokasi titik pengamatan.

### **Constraints**

> Jangan mengolah ulang data dan jangan membuat nilai baru. Gunakan nilai, satuan, dan legenda yang terdapat pada hasil yang diberikan.

### **Output**

> Jelaskan wilayah yang mengalami deformasi, arah perubahannya, dan bagaimana perubahan tersebut berlangsung terhadap waktu, misalnya meningkat, menurun, atau relatif stabil.

### **Quality Criteria**

> Interpretasi harus sesuai dengan pola pada peta dan grafik serta tidak memberikan kesimpulan yang melebihi informasi yang tersedia.

### **Verification**

> Bandingkan hasil interpretasi pada grafik dengan velocity map dan data pembanding jika tersedia.





## **4️. STUDI KASUS SEISMIK REFRAKSI**

### **Deskripsi**

Pada studi kasus ini, hasil seismik refraksi digunakan untuk memahami **susunan lapisan bawah permukaan** berdasarkan informasi kecepatan dan kedalaman.



### **Role**

> Anda berperan sebagai pendamping dalam membantu memahami hasil analisis seismik refraksi.

### **Task**

> Bantu kita menginterpretasikan hasil seismik refraksi untuk mengetahui gambaran lapisan bawah permukaan.

### **Context**

> Seismik refraksi menggunakan waktu tiba gelombang untuk memberikan informasi mengenai kecepatan dan kedalaman lapisan bawah permukaan.

### **Input**

> Kita akan memberikan kurva travel time, model kecepatan, kedalaman lapisan, dan informasi geologi jika tersedia.

### **Constraints**

> Jangan menghitung ulang atau membuat nilai baru. Jangan langsung menentukan jenis batuan hanya berdasarkan nilai kecepatan.

### **Output**

> Jelaskan jumlah lapisan, perbedaan kecepatannya, kedalaman batas lapisan, serta kemungkinan material penyusunnya.

### **Quality Criteria**

> Interpretasi harus sesuai dengan model kecepatan yang diberikan dan mempertimbangkan informasi geologi yang tersedia.

### **Verification**

> Periksa apakah interpretasi lapisan sudah sesuai dengan model kecepatan dan data geologi pendukung.





## **5️. STUDI KASUS GRAVITASI**

### **Deskripsi**

Pada studi kasus ini, kita menggunakan **peta Bouguer anomaly** untuk membantu memahami variasi densitas dan kemungkinan struktur bawah permukaan.



### **Role**

> Anda berperan sebagai pendamping dalam membantu memahami dan menjelaskan hasil analisis gravitasi.

### **Task**

> Bantu kita menginterpretasikan pola anomali gravitasi untuk melihat kemungkinan perubahan densitas dan struktur bawah permukaan.

### **Context**

> Metode gravitasi dapat menunjukkan adanya perbedaan densitas material bawah permukaan melalui perubahan nilai medan gravitasi.

### **Input**

> Kita akan memberikan peta Bouguer anomaly, nilai anomali, lokasi penelitian, dan peta geologi jika tersedia.

### **Constraints**

> Jangan mengolah ulang data dan jangan membuat nilai baru. Jangan langsung menentukan jenis batuan hanya berdasarkan nilai anomali gravitasi.

### **Output**

> Jelaskan zona anomali positif dan negatif, kemungkinan perubahan densitas, serta kemungkinan adanya struktur geologi.

### **Quality Criteria**

> Interpretasi harus berdasarkan pola anomali yang terlihat dan didukung oleh informasi geologi jika tersedia.

### **Verification**

> Periksa kembali apakah interpretasi sesuai dengan peta gravitasi dan bandingkan dengan data geologi atau data geofisika lain jika tersedia.
